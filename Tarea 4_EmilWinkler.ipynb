# Primera parte. Ecuación de Difusión en 1D


## Pregunta 1

La ecuación de difusión que resolviste en el ejercicio anterior era: 

$$\frac{\partial P(x,t)}{\partial t} = D \frac{\partial^2 P(x,t)}{\partial x^2}$$

Encuentra la transformada de Fourier, de ambos lados de la ecuación. Recuerda que la transformada de Fourier está dada por:

$$\hat{P}(k,t) = \int_{-\infty}^{\infty} P(x,t) e^{-i2\pi kx} dx$$

[nota que estamos transformando del espacio de al espacio de las frecuencias espaciales:]

Para poder realizar este método tenemos que ver como es la transformada de Fourier para nuestros términmos que estan siendo derivados. Empezaremos con la derivada de nuestro lado izquierdo (la derivada temporal), y luego seguiremos con el lado derecho (segund derivada espacial).

### Lado izquierdo/Derivada temporal

$$\mathcal{F} \left\{ \frac{\partial P}{\partial t} \right\} = \int_{-\infty}^{\infty} \frac{\partial P}{\partial t} e^{-i2\pi kx} dx$$

Como la transformada de Fourier no esta afectando directamente el tiempo de nuestra función $P(x,t)$, podemos separar la derivada temporal de nuestro integrando y por lo tanto llegar a una expresión en términos de nuestra función $\hat{P}(k,t)$ en el dominio de frecuencias.

$$  \mathcal{F} \left\{ \frac{\partial P}{\partial t} \right\} = \frac{\partial }{\partial t} \int_{-\infty}^{\infty} P e^{-i2\pi kx}$$

$$  \mathcal{F} \left\{ \frac{\partial P}{\partial t} \right\} = \frac{\partial \hat{P}(k,t) }{\partial t}

### Lado derecho/Segunda derivada espacial

$$\mathcal{F} \left\{ D \frac{\partial^2 P}{\partial x^2} \right\} = \int_{-\infty}^{\infty}  D \frac{\partial^2 P}{\partial x^2}  e^{-i2\pi kx} dx$$

En este caso no podemos sacar la derivada directamente por la dependencia que tiene con el espacio ($x$), por lo que tenemos que analizar la formula en la que se realiza la transformada de una derivada. 

De forma generalizada tenemos una función lo suficientemente suave $f(x)$ y su derivada $f'(x)$. La transformada de su derivada se define entonces como:

$$ \mathcal{F} \left\{f'(x) \right\} = \int_{-\infty}^{\infty} f'(x)  e^{-i2\pi kx} dx$$

Ahora utilizaremos integración por partes para poder operar la integral con los parámetros $u = e^{-i2\pi kx}$ y $dv = f'(x)$

$$du = -i2\pi k e^{-i2\pi kx} $$

$$v = f(x)$$

$$ \mathcal{F} \left\{f'(x) \right\} =  \left[ f(x) e^{-i2\pi kx} \right]_{-\infty}^{\infty} - \int_{-\infty}^{\infty} f(x)  (-i2\pi k) e^{-i2\pi kx} dx$$

Como nuestra $f(x)$ es una función suave que representa un significado físico, la función que se evalua en $\infty$ y -$\infty$ se vuelve 0. 

$$ \mathcal{F} \left\{f'(x) \right\} = 0 + i2\pi k \int_{-\infty}^{\infty} f(x) e^{-i2\pi kx} dx$$


$$ \mathcal{F} \left\{f'(x) \right\} = i2\pi k \int_{-\infty}^{\infty} f(x) e^{-i2\pi kx} dx =  i2\pi k \hat{P}(k,t)$$

Si nos damos cuenta,  al final nos queda la misma transformada por un factor de $i2 \pi kx$. Si volvemos a derivar esto para obtener la segunda derivada nos damos cuenta que lo que sucedera es que se añadira nuevamente el factor $i2 \pi kx$. Por ende:


$$ \mathcal{F} \left\{f^n(x) \right\} =  (i2\pi k)^n \hat{f}(k,t) $$

Sabiendo esto podemos determinar nuestra segunda derivada espacial de $\hat{P}$.

$$\mathcal{F} \left\{ D \frac{\partial^2 P}{\partial x^2} \right\} = D (i2 \pi k)^2  \hat{P}(k,t)$$

$$ \mathcal{F} \left\{ D \frac{\partial^2 P}{\partial x^2} \right\} = -4D\pi^2k^2  \hat{P}(k,t)$$

## Pregunta 2

Demuestra que la transformada de la función $\hat{P}(k,t)$ está dada por una ecuación diferencial ordinaria. Escribe esta ecuación. 

Si juntamos las dos partes de la ecuación, obtenemos la relación directa de la derivada temporal con la distribución de probabilidad transofrmada.

$$ \frac{\partial \hat{P}(k,t) }{\partial t} = -4D\pi^2k^2  \hat{P}(k,t)$$

Sin embargo, como tenemos  ahora en $\hat{P}(k,t)$ una constante de frecuencia $k$. Al derivar respecto al tiempo, la variable $k$ se comporta como un parámetro fijo. Con ello, para un valor de $k$, la función $\hat{P}(k,t)$ solo puede variar en el tiempo y por definición la deriada parcial se comporta como total. Por lo mismo, ahora tenemos una **ecuación diferencial ordinaria**, donde tenemos $k$ como variable independiente y $t$ como la función que cambia. Después de la transformada, vivimos en una sola dimensión de cambio y por ende nuestro problema se simplifica significativamente 

$$ \frac{d \hat{P}(k,t) }{d t} = -4D\pi^2k^2  \hat{P}(k,t)$$


## Pregunta 3

Suponiendo que la condición inicial de la ecuación de difusión es es la función $P(x,0)$  ¿Cuál seria la forma adecuada de esta función para esta ecuación diferencial ordinaria?

Las ecuaciones diferenciales ordinarias que dependen del tiempo necesitan de una condición inicial en un tiempo dado. En el caso de este problema de difusión tenemos una condición inicial en el instante $t = 0$. Con esto en mente, tenemos que encontrar el análogo en nuestra ODE. 

Como en la ecuación diferencial ordinaria estamos trabjando con la transformada de la función $P(x,t)$, la cual vive en el dominio de las frecuencias $k$, nuestra condición inicial conlleva el mismo principio del tiempo cero ya que a pesar de vivir en un dominio completamente distinto, el tiempo se mantiene invariante ante la transformada. 

Consecuentemente,

$$ \hat{P}(k,0) =  \int_{-\infty}^{\infty} P(x,0) e^{-i2\pi kx} dx $$

## Pregunta 4


En este caso, la ecuación diferencial ordinaria se puede resolver de forma analítica. Encuentra la solución completa, incluyendo las condiciones iniciales.

Para poder resolver este problema de forma análitica tenemos que partir de nuestro resultado de las preguntas 2 y 3. 

Tenemos la ecuación diferencial ordinaria:

$$ \frac{d \hat{P}(k,t) }{d t} = -4D\pi^2k^2  \hat{P}(k,t)$$

Para resolverla aplicaremos el método de separación de variables que nos permite sacar las soluciones de EDOs de manera eficaz. Como primer paso, separamos nuestras variables para tener la función $\hat{P}(k,t)$ de un lado y el tiempo del otro. 

$$ \frac{d \hat{P}(k,t) }{\hat{P}(k,t)} = -4D\pi^2k^2  \hat{P}(k,t) dt$$

Una vez hecho esto es posible integrar ambas partes para quitarnos el diferencial. 
 
$$ \int \frac{d \hat{P}(k,t) }{\hat{P}(k,t)} = \int -4D\pi^2k^2  \hat{P}(k,t) dt$$

Al ser una integral indefinida aparece una constante en cada lado que podemos agrupar a la derecha llamandole $C$

$$ \ln (\hat{P}(k,t)) = -4D \pi^2 k^2 t + C $$

Ahora para quitarnos el logaritmo natural vamos a utilizar un exponencial en cada lado de la ecuación.

$$ e^{\ln (\hat{P}(k,t))}  = e^{-4D \pi^2 k^2 t + C} $$

Como se esta sumando un exponente podemos separar la exponencial en dos y tratarlas por separado.

$$ e^{\ln (\hat{P}(k,t))}  = e^C ·e^{-4D \pi^2 k^2 t}  $$

Como $e^C$ es un número $\mathbb{R}$, podemos considerarlo como otra constante $A$. Con esta consideración habremos encontrado una expresión generalizada de la familia de soluciones. 

$$ \hat{P}(k,t) = Ae^{-4D \pi^2 k^2 t}  $$

Sin embargo, nosotros podemos utilizar nuestra condición inicial para poder darle un valor a $A$ y un significado más específico a la ecuación en relación de nuestro problema. De esta forma, consideraremos la condición inicial conocida del tiempo inicial $t = 0$, donde nuestra función $ \hat{P}(k,t)$ vale $ \hat{P}(k,0)$. 

$$ \hat{P}(k,0) = Ae^0 $$

$$ \hat{P}(k,0) = A $$

Con esto, tenemos la solución de nuestra ODE en nuestro dominio de frecuencias de forma completa. 

$$  \hat{P}(k,t) = \hat{P}(k,0)e^{-4D \pi^2 k^2t}


## Pregunta 5

Programa un método que evalúe de forma numérica esta solución, y luego utilice la transformada inversa para encontrar el valor de la función discretizada $P(x_i, t_i)$. Recuerda que puedes utilizar la transformada rápida de Fourier (fft) para calcular la transformada de Fourier de una función discreta.




```python
import numpy as np
import matplotlib.pyplot as plt

def fft_difusion(x, P_0, D, t):
    N = len(x)

    # Definimos el paso espacial
    dx = x[1] - x[0]

    # Creamos el vector de frecuencias para la transformada
    k = np.fft.fftfreq(N, d=dx) 

    # Factor de decaimiento
    decaimiento = np.exp(-4 * D * np.pi**2 * k**2 * t)

    # Transformada de Fourier de la condición inicial
    P_0_fft = np.fft.fft(P_0)

    P_gorrito = P_0_fft * decaimiento

    # Transformada inversa
    P_xt = np.fft.ifft(P_gorrito)

    return P_xt.real

# Parámetros físicos y de malla
D_coef = 1.0  # Coeficiente de difusión
L =  100.0  # Longitud del dominio
N_puntos = 512 # Número de puntos en la malla espacial
x_grid = np.linspace(-L/2, L/2, N_puntos, endpoint=False) # Partimos el dominio en N_puntos

# Definimos una condición inicial P(x, 0): Una campana de Gauss estrecha en el origen
P_x0 = np.exp(-(x_grid**2) / 0.5)

# Normalizamos para que el área bajo la curva sea 1
area = np.sum(P_x0) * (x_grid[1] - x_grid[0])
P_x0 /= area
```

```python
# Tiempos a evaluar
tiempos = [20, 40, 60, 100]

# Graficamos los resultados
plt.figure(figsize=(10, 6))

# Calculamos y gráficamos la solución para cada tiempo.
for ti in tiempos:
    if ti == 0:
        P_xt = P_x0
    else:
        P_xt = fft_difusion(x_grid, P_x0, D_coef, ti)
        
    plt.plot(x_grid, P_xt, label=f'Tiempo t = {ti}')

plt.title('Solución de la Ecuación de Difusión usando FFT')
plt.xlabel('Posición (x)')
plt.ylabel('Probabilidad P(x,t)')
plt.legend()
plt.grid(True)
plt.show()
```

# Parte 2 Ecuación no lineal de Burgers con viscosidad 

## Pregunta 7

La ecuación no lineal de Burgers es una ecuación que ocurre con frecuencia en fenómenos de transporte forzado, y tiene la característica de generar ondas de choque. La forma viscosa de la ecuación de Burgers es

$$ \frac{\partial u}{\partial t} + \frac{1}{2}\frac{\partial(u^2)}{\partial x} = D \frac{\partial^2 u}{\partial x^2} $$
Al igual que en el procedimiento anterior, transforma ambos lados de la ecuación, y demuestra que la transformada 
$\hat{u}(k,t)$ está descrita por una ecuación diferencial ordinaria. 

Escribe esta ecuación en la forma $\frac{d \hat{u}(k,t)}{dt} = f(t,\hat{u}(k,t)) $ en donde $f$ es una función arbitraria. 




### Lado derecho

Utilizando la demostración de las derivadas en el espacio frecuencias, podemos sacar la segunda derivada espacial de $\hat{u}(k,t)$ como una sencilla multiplicación. 


$$ \mathcal{F} \left\{ D \frac{\partial^2 u}{\partial x^2} \right\} = -4D\pi^2k^2  \hat{u}(k,t)$$

### Lado izquierdo

$$  \mathcal{F} \left\{ \frac{\partial u}{\partial t} \right\} = \frac{d \hat{u}(k,t) }{dt} $$

Sabiendo que la dependencia del tiempo se vuelve total y como menciona el problema, para nuetro primer término de $u$ definimos una función arbitraria. 

$$ \frac{d \hat{u}(k,t) }{dt} = f(t,\hat{u}(k,t)) $$
en donde $f$ es una función arbitraria. 

Finalmente, transformamos la función no lineal del problema

$$\mathcal{F} \left\{\frac{1}{2}\frac{\partial(u^2)}{\partial x} \right\}$$

Como sabemos, al tener una derivada espacial podemos utilizar la propiedad y sacar el factor del exponencial y multiplicar por la transformada de nuestra función $u^2$. No obstante, no podemos decir de manera directa que $\mathcal{F} \left\{u^2 \right\} = (\hat{u})^2 $. De hecho, gracias al teorema de convolución, esto es realmente la multiplicación de la función por si misma en el espacio de frecuencias. Por esta razón dejaremos expresado esta transformada como $\mathcal{F} \left\{u^2 \right\} $.

$$\mathcal{F} \left\{\frac{1}{2}\frac{\partial(u^2)}{\partial x} \right\} = \frac{1}{2}(2i \pi k) \mathcal{F} \left\{u^2 \right\}$$

$$\mathcal{F} \left\{\frac{1}{2}\frac{\partial(u^2)}{\partial x} \right\} = i \pi k \mathcal{F} \left\{u^2 \right\}$$

Después de haber transformado ambos lados de nuestra ecuación la podemos construir en el dominio frecuencial, 

$$f(t,\hat{u}(k,t)) = -4D\pi^2k^2  \hat{u}(k,t) - i \pi k \mathcal{F} \left\{u^2 \right\}$$

## Pregunta 8

Programa un algoritmo para resolver la ecuación que describe la función 
$\hat{u}(k,t)$. Utiliza la transformada inversa para obtener el valor de la función 
$u(x,t)$. 

En este caso como tenemos el factor de $u^2$ tenemos que recalcular su valor en cada paso del tiempo. Por ende: 


```python
import numpy as np
import matplotlib.pyplot as plt

def derivada_temporal(u_0_fft, k, D):
    factor_difusion = -4 * D * np.pi**2 * k**2
    # Definimos primer termino de nuestra ecuación
    termino_difusion = factor_difusion * u_0_fft
    
    u_real = np.fft.ifft(u_0_fft).real
    u_cuadrada = u_real **2

    # Transformamos el termino no lineal
    u_cuadrada_fft = np.fft.fft(u_cuadrada)

    # Definimos termino no lineal 
    termino_no_lineal = -1j * np.pi * k * u_cuadrada_fft

    # Definimos la derivada temmporal en el espacio de la ecuacion
    return termino_difusion + termino_no_lineal
    

def fft_burgers(x, u_0, D, t, dt):
    N = len(x)
    
    # Definimos el paso espacial
    dx = x[1] - x[0]

    # Generamos el vector de frecuencias para la transformada 
    k = np.fft.fftfreq(N, d=dx)

    # Transformada del la condicion inicial del espacio
    u_0_fft = np.fft.fft(u_0)
    
    # Discretizamos el tiempo
    pasos = int(t/dt)

    u_gorrito = u_0_fft

    # Iteramos haciendo RK4 para cada paso del tiempo
    for _ in range(pasos):
        k1 = derivada_temporal(u_gorrito, k, D)
        k2 = derivada_temporal(u_gorrito + 0.5 * dt * k1, k, D)
        k3 = derivada_temporal(u_gorrito + 0.5 * dt * k2, k, D)
        k4 = derivada_temporal(u_gorrito + dt * k3, k, D)

        u_gorrito = u_gorrito + (dt / 6.0) * (k1 + 2*k2 + 2*k3 + k4)
    
    return np.fft.ifft(u_gorrito).real
```

## Pregunta 9

Utiliza las mismas condiciones que para la ecuación de difusión. Grafica el valor de la función después de transcurridos 20, 40, 60 y 100 segundos. Compara tus resultados

```python
# Parámetros físicos y de malla
D_coef = 1.0  # Coeficiente de difusión
L =  100.0  # Longitud del dominio
N_puntos = 512 # Número de puntos en la malla espacial
x_grid = np.linspace(-L/2, L/2, N_puntos, endpoint=False) # Partimos el dominio en N_puntos
dx = x_grid[1] - x_grid[0]

# Definimos una condición inicial para u.
# Una campana de gauss estrecada en el origen
u_x0 = np.exp(-(x_grid)**2 / 0.5)
area = np.sum(u_x0) * dx
u_x0 /= area

# Tiempos a evaluar 
tiempos = [20,40,60,100]
dt = 0.01

# Graficamos el método espectral

plt.figure(figsize=(10,6))

for ti in tiempos:
    if ti == 0:
        u_xt = u_x0
    else:
        u_xt = fft_burgers(x_grid, u_x0, D_coef, ti, dt)
    plt.plot(x_grid, u_xt, label=f'Tiempo t = {ti}')

plt.title('Solución de la Ecuación de Burgers usando FFT')
plt.xlabel('Posición (x)')
plt.ylabel('Amplitud u(x,t)')
plt.legend()
plt.grid(True)
plt.xlim(-50,50)
plt.show()
```
